# Benchmark figures

Run the cells in order after setting `CSV_PATH` to the evaluator’s `paper_metrics.csv`.
The CSV supplies the models, conditions and statistics; nothing is recalculated here.
Missing statistics remain N/A. E5 outcomes are displayed only when E5 is present.

Each cell calls a function in `plot_paper_results.py`, also used by the command line.
The last cell optionally saves PDF, SVG and PNG files. Saved outputs are not included in this notebook.


```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents)
            if (p / "notebooks/plot_paper_results.py").is_file())
sys.path.insert(0, str(ROOT / "notebooks"))
import plot_paper_results as plots

CSV_PATH = ROOT / "reports/evaluation/paper_metrics.csv"
df = plots.load_results([CSV_PATH])
```

```python
with plt.rc_context():
    fig = plots.plot_radar(df)
    if fig is not None:
        display(fig)
        plt.close(fig)
```

```python
with plt.rc_context():
    fig = plots.plot_scatter(df)
    if fig is not None:
        display(fig)
        plt.close(fig)
```

```python
with plt.rc_context():
    fig = plots.plot_heatmap(df)
    if fig is not None:
        display(fig)
        plt.close(fig)
```

```python
with plt.rc_context():
    fig = plots.plot_identification(df)
    if fig is not None:
        display(fig)
        plt.close(fig)
```

```python
with plt.rc_context():
    fig = plots.plot_completion_gap(df)
    if fig is not None:
        display(fig)
        plt.close(fig)
```

```python
# Set to True to export all applicable figures and their input manifest.
SAVE_FIGURES = False
if SAVE_FIGURES:
    files = plots.render_all(df, ROOT / "reports/figures")
    print(f"Saved {len(files)} files to reports/figures")
```
